```python
sns.countplot(x='species', data=df)

plt.title('Distribution of Penguin Species')
plt.xlabel('Species')
plt.ylabel('Count')
plt.show()
```

```python
df['species'].value_counts()
```

```python
df['species'].value_counts().idxmax()
```

```python
island_percentage = (df['island'].value_counts(normalize=True) * 100).round(2)

print("Percentage Distribution of Islands:")
print(island_percentage)
```

```python
sex_counts = df['sex'].value_counts()

print(sex_counts)

plt.figure(figsize=(6,6))
plt.pie(
    sex_counts,
    labels=sex_counts.index,
    autopct='%1.1f%%'
)

plt.title('Gender Distribution of Penguins')
plt.show()
```
