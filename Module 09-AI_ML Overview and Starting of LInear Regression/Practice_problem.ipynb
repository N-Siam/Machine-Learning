## Simple Linear Regression
1. Simple Linear Regression কী predict করে?
2. Independent Variable এবং Dependent Variable এর পার্থক্য লিখো।
3. y = mx + b equation এ m এবং b কী বোঝায়?
4. Best Fit Line বলতে কী বুঝায়?
5. Linear Regression এর একটি real-life use case লিখো।


## Simple Linear Regression

1. **Simple Linear Regression:** Predicts a dependent variable (Y) using one independent variable (X).

2. **Independent Variable:** Input used for prediction.  
   **Dependent Variable:** Output being predicted.

3. **m:** Slope or rate of change.  
   **b:** Y-value when X = 0.

4. **Best Fit Line:** A line that best represents the relationship between X and Y with minimum error.

5. **Use Case:** Predicting house prices based on house size.

## Cost Function
1. Cost Function কেন দরকার?
2. Prediction Error কী?
3. Model error কমানো কেন গুরুত্বপূর্ণ?
4. Large error হলে Cost Function এর value কেমন হবে?
5. Cost কমানোর process এর নাম কী?


## Cost Function

1. **Cost Function:** Measures how much error a model makes.

2. **Prediction Error:** The difference between the actual value and predicted value.

3. **Why reduce error?** To make the model more accurate.

4. **Large Error:** The Cost Function value will be **large**.

5. **Cost Reduction Process:** *Minimizing the Cost Function*.

### Task 3

```python
# Python list ব্যবহার করে ৫টি house price store করো
house_prices = [150000, 180000, 210000, 250000, 300000]
print(house_prices)
```

### Task 4

```python
# matplotlib ব্যবহার করে একটি straight line plot করো

import matplotlib.pyplot as plt

house_sizes = [1000, 1200, 1400, 1600, 1800]
house_prices = [290000, 350000, 410000, 480000, 550000]

plt.plot(house_sizes, house_prices, marker='o', color='red')
plt.title("Straight Line")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

### Task 5

```python
# sklearn ব্যবহার করে একটি simple linear regression model তৈরি করো

# House sizes
house_sizes = [[1000], [1200], [1400], [1600], [1800]]

# Create model
from sklearn.linear_model import LinearRegression

model = LinearRegression()

# Train model
model.fit(house_sizes, house_prices)

# Predict price for 1500 sq ft
prediction = model.predict([[1500]])

print(prediction)
```
